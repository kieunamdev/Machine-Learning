### Gradient Descent

```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

## VD1: f(x)=x**2 + 5sin(x)
  

```python
def grad(x):
    return 2*x + 5*np.cos(x)

def cost(x):
    return x**2 + 5*np.sin(x)

def myGD1(eta, x0):
    x = [x0]
    for it in range(100):
        x_new = x[-1] - eta*grad(x[-1]) 
        if abs(grad(x_new)) < 1e-3:
            break
        x.append(x_new)
    return (x, it)

(x1, it1) = myGD1(0.1, -5)
(x2, it2) = myGD1(0.1, 5)
print('Solution x1 = %f, cost = %f, obtained after %d iterations'%(x1[-1], cost(x1[-1]), it1))
print('Solution x2 = %f, cost = %f, obtained after %d iterations'%(x2[-1], cost(x2[-1]), it2))
```

## VD2: f(x) = x**2 -2

```python
def grad(x):
    return 2*x

def cost(x):
    return x**2 -2

def myGD2(eta, x0):
    x = [x0]
    for it in range(100):
        x_new = x[-1] - eta*grad(x[-1])
        if abs(grad(x[-1])) < 1e-3:
            break
        x.append(x_new)
    return (x, it)

(x1, it1) = myGD2(0.1, -5)
(x2, it2) = myGD2(0.1, 5)

print('Solution x1 = %f, cost = %f, obtained after %d iterations'%(x1[-1], cost(x1[-1]), it1))
print('Solution x2 = %f, cost = %f, obtained after %d iterations'%(x2[-1], cost(x2[-1]), it2))
```

## VD3:f(x) = (1/3)*x**3 - x

```python
def grad(x):
    return x**2 -1

def cost(x):
    return (1/3)*x**3 - x

def myGD3(eta, x0):
    x = [x0]
    for it in range(100):
        x_new = x[-1] - eta*grad(x[-1])
        if abs(grad(x_new)) < 1e-3:
            break
        x.append(x_new)
    return (x, it)

(x1, it1) = myGD3(0.1, -0.5)
(x2, it2) = myGD3(0.1, 5)

print('Solution x1 = %f, cost = %f, obtained after %d iterations'%(x1[-1], cost(x1[-1]), it1))
print('Solution x2 = %f, cost = %f, obtained after %d iterations'%(x2[-1], cost(x2[-1]), it2))
```
