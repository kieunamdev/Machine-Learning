```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Step1: Read DataSet
data = pd.read_csv('../data/gia_nha.csv', index_col='ID_nha')
data.head()
```

```python
#Step2: Extract data
features = ['Dien_tich', 'So_phong']
X = data[features]
y = data['Gia_ty']
```

```python
#Step3: Split Train/Test
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, train_size=0.8, test_size=0.2, random_state=42)
```

```python
# Step 4: So sanh do phuc tap cua mo hinh

degrees = [1, 3, 5, 7, 9]

results = []
```

```python
# Step 5: Train cac mo hinh voi cac degree khac nhau

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

for degree in degrees:

    # Tao polynomial features
    poly = PolynomialFeatures(
        degree=degree,
        include_bias=False
    )

    # Bien doi du lieu Train va Test
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)

    # Tao va train mo hinh
    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    # Du doan
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)

    # Tinh MSE
    train_mse = np.mean((y_train - y_train_pred) ** 2)
    test_mse = np.mean((y_test - y_test_pred) ** 2)

    
    results.append([
        degree,
        train_mse,
        test_mse
    ])
```

```python
# Step 6: Ket qua

results_df = pd.DataFrame(
    results,
    columns=['Degree', 'Train MSE', 'Test MSE']
)

results_df
```
